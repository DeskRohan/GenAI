```python
import gensim.downloader as api
from scipy.spatial.distance import cosine

model = api.load("word2vec-google-news-300")

print("First 10 dimensions of king:")
print(model["king"][:10])

print("\nTop 10 similar words to king:")
for w, s in model.most_similar("king", topn=10):
    print(w, round(s, 4))

print("\nking - man + woman =")
print(model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=1
))

print("\nparis + italy - france =")
print(model.most_similar(
    positive=["paris", "italy"],
    negative=["france"],
    topn=5
))

print("\nwalking + swimming - walk =")
print(model.most_similar(
    positive=["walking", "swimming"],
    negative=["walk"],
    topn=5
))

sim = 1 - cosine(model["king"], model["queen"])
print("\nKing-Queen Similarity:", round(sim, 4))
```
